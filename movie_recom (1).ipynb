```python
import ast
def convert_genres(obj):
    L = []
    for i in ast.literal_eval(obj):
        L.append(i['name'])
    return " ".join(L)
```

```python
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
import re
```

```python
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()
```

```python
def preprocess(text):
  #convert to lower case
  text = str(text).lower()

  #remove non alphabetic chars
  text = re.sub(r'[^a-zA-Z\s]', '', text)

  #tokenization
  words = text.split()

  #remove stopwords e.g is,a
  words = [word for word in words if word not in stop_words]

  #lemmatization e.g running or runs comes to root form run
  words = [lemmatizer.lemmatize(word) for word in words]

  return " ".join(words)

# Apply preprocessing to the 'features' column
df['features'] = df['features'].apply(preprocess)
```

```python
df['features'].head()
```

```python

# Drop rows with duplicate titles so each title is unique
df = df.drop_duplicates(subset=['title']).reset_index(drop=True)

#find row number of a movie using its title
indices = pd.Series(df.index, index=df['title'])
```

```python
indices
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
movies=TfidfVectorizer(
    max_features=50000,
    ngram_range=(1,2),
    stop_words='english',
    min_df=2,
    max_df=0.9
)
# min_df removes noise words that appear in very few documents ,max_df removes words very common
# ngram word and word pairs
```

```python
movies_matrix = movies.fit_transform(df['features'])
```

```python
movies_matrix
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
def recommend(movie, n=10):
    if movie not in indices:
        return "Movie not found"

    idx = indices[movie]

   # Compute similarities and compares movie similarity to other movies in dataset
    similarity_scores = cosine_similarity(
        movies_matrix[idx],
        movies_matrix
    ).flatten()

    # Get top matches by argsort[::-1] returns [[indices sorted in reverse order of similarity]] [1:n+1] skips 0th index movie
    similar_indices = similarity_scores.argsort()[::-1][1:n+1]

    # Return the titles from the dataframe
    return df['title'].iloc[similar_indices]
```

```python
def recommend_and_scores(movie, n=10):
    if movie not in indices:
        return "Movie not found"

    idx = indices[movie]

    # Compute similarity scores
    similarity_scores = cosine_similarity(
        movies_matrix[idx],
        movies_matrix
    ).flatten()

    # Get top matches excluding the movie itself
    similar_indices = similarity_scores.argsort()[::-1][1:n+1]

    # Create a DataFrame with titles and their respective scores
    recommendations = pd.DataFrame({
        'title': df['title'].iloc[similar_indices].values,
        'similarity_score': similarity_scores[similar_indices]
    })

    return recommendations
```

```python
recommend_and_scores('Avatar',10)
```

```python
recommend('Batman',10)
```

```python
recommend('Avatar',5)
```
